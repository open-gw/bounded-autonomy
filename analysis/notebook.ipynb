# Paper 1 analysis (pre-registration)

Reads five `result.json` files per mode and emits the three manuscript tables plus the Q2 figure. Metric definitions are frozen in `docs/study-design.md`; this notebook does not redefine them.

Paste the Markdown tables into the manuscript. Fill the `\val{}` slots from the same numbers.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "analysis" else Path.cwd()
sys.path.insert(0, str(ROOT / "analysis"))
from tables import (
    FIXTURE_RESULTS,
    DEFAULT_RESULTS,
    emit_markdown,
    load_results,
    write_q2_figure,
)

results_dir = DEFAULT_RESULTS
if not list(results_dir.glob("**/result.json")):
    results_dir = FIXTURE_RESULTS
results = load_results(results_dir)
print(f"loaded {len(results)} results from {results_dir}")
assert len(results) == 10, "Paper 1 expects five seeds × two modes"
assert {r['mode'] for r in results} == {"flat", "full"}
```

## Reach

```python
from tables import table_reach, table_rollback, table_overhead

print(table_reach(results))
```

## Rollback

```python
print(table_rollback(results))
```

## Overhead

```python
print(table_overhead(results))
```

## Q2 figures

```python
out = ROOT / "analysis" / "output"
path = write_q2_figure(results, out)
md = emit_markdown(results)
(out / "tables.md").write_text(md)
print(md)
print("figure:", path)
```
